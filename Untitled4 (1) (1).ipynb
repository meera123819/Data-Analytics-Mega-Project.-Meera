```python
import seaborn as sns
```

```python
num_cols=df.select_dtypes(include='number'
                ).columns
plt.figure(figsize=(8,5))
plt.hist(df[num_cols[0]],
         bins=20)
plt.title(f"Distribution of{num_cols[0]}")
plt.xlabel(num_cols[0])
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df[num_cols[0]])
plt.title(f"Box Plot of {num_cols[0]}")
plt.show()
```

```python


cat_cols= df.select_dtypes(include='object').columns
plt.figure(figsize=(7,7))
df[cat_cols[0]].value_counts().head(6).plot(
        kind='pie',
        autopct='%1.1f%%'
    )
plt.title(f"Distribution of {cat_cols[0]}")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=df[num_cols[0]],
    y=df[num_cols[1]]
)
plt.title(f"{num_cols[0]} vs {num_cols[1]}")
plt.xlabel(num_cols[0])
plt.ylabel(num_cols[1])
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df[num_cols[0]].head(30),marker='0')
plt.title(f"Trend of{num_cols[0]}")
plt.xlabel("Record")
plt.ylabel(num_cols[0])
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x=df[cat_cols[0]])
plt.title(f"Count of{cat_cols[0]}")
plt.xlabel(cat_cols[0])
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(
    df[num_cols].corr(),
    annot=True,
    cmap='coolwarm'
    )
plt.title("Correlation Heatmap")
plt.show()
```
